## Phase 3: Implementation (CS229 Python Notebook)

- Data Preparation
  - Load housing data or generate synthetic linear data with noise
  - Add intercept column (x_0 = 1) to design matrix X

- Algorithm 1: Closed-Form (Normal Equation)
  - Compute θ using np.linalg.solve(X.T @ X, X.T @ y)
  - Print parameters and plot fitted line

- Algorithm 2: Batch Gradient Descent
  - Initialize θ to zeros
  - Loop over full dataset: ∇J = (X.T @ (X @ θ - y)) / n
  - Update θ := θ - α * ∇J for fixed iterations
  - Plot cost history J(θ) over iterations

- Algorithm 3: Stochastic Gradient Descent
  - Shuffle data at start of each epoch
  - Loop per sample: θ := θ - α * (h(x^(i)) - y^(i)) * x^(i)
  - Track moving average of cost to show oscillation

- Evaluation
  - Compute RMSE on training set for all three methods
  - Verify all three yield approximately the same final θ
